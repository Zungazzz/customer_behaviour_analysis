```python
# Necessary imports
import pandas as pd

# # Reading dataset from google drive
# file_path = '/content/drive/MyDrive/data/customer_shopping_behavior.csv'

# Reading dataset locally
file_path = r'C:\Users\Loojah Computers\Documents\GitHub\customer_behaviour_analysis\data\raw\customer_shopping_behavior.csv'

df = pd.read_csv(file_path)
# First 5 rows of dataset
df.head()
```

```python
# # Check dataset shape, column data types (dtypes), missing values, and memory usage
df.info()
```

```python
# Summarize every column (numeric stats + text/categorical stats; NaN where a metric doesn't apply)
df.describe(include='all')
```

```python
# Count for null rows in each columns
df.isnull().sum()
```

```python
# Filling null rows with mean of respective Category
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Standarizing the columms names
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns= {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# Create a column age_group
labels= ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# Create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Every 3 Months' : 90,
    'Annually' : 365
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
# Checking if 'discount_applied' and 'promo_code_used' have same values
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
from sqlalchemy import create_engine, text

# Step 1: Connect to PostgreSQL
username = 'postgres'
password = '1234'
host = 'localhost'
port = 5432
database = 'customer_shopping_behaviour'

engine = create_engine(
    f'postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}'
)

# Step 2: Load DataFrame into PostgreSQL
table_name = 'shopping_behavior'

with engine.begin() as conn:
  df.to_sql(table_name, con=conn, if_exists='replace', index=False)
  row_count = conn.execute(text(f'SELECT COUNT(*) FROM {table_name}')).scalar()

print(
    f"Data successfully loaded ({row_count} rows) into table '{table_name}'"
    f" in database '{database}'."
)
```
